```python
# for negative values

def Reverse_Number(n) :

    Int_Min , Int_Max = -2**31 , 2**31-1

    if n < 0 :
        sign = -1

        n = abs(n)
        n = str(n)
        n = n[::-1]
        n = int(n)
        n = n * sign

    else :
        n = str(n)
        n = n[::-1]
        n = int(n)

    if n < Int_Min or n > Int_Max :   # the problem expect a 32-bit integer result 
        return 0                      # if the n value is outise the (-2**31) or (2**31-1) it will return ZERO [0]
    
    return(n)

Reverse_Number(-123)
```

```python
s = "-87"


INT_MIN = -2**31
INT_MAX = 2**31 - 1

i = 0
n = len(s)

# Step 1: Ignore leading spaces
while i < n and s[i] == " ":
    i += 1

# Step 2: Determine the sign
sign = 1

if i < n and s[i] == "-":
    sign = -1
    i += 1
elif i < n and s[i] == "+":
    i += 1

# Step 3: Read consecutive digits
num = 0

while i < n and s[i].isdigit():
    digit = int(s[i])
    num = num * 10 + digit

    # Check overflow while building the number
    if sign == 1 and num > INT_MAX:
        print(INT_MAX)

    if sign == -1 and -num < INT_MIN:
        print(INT_MIN)

    i += 1

# Step 4: Apply sign
sign = sign * num
print(sign)

# Hence its an Integer
print(sign + 1)
```
