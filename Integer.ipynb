```python
s = "-87"


INT_MIN = -2**31
INT_MAX = 2**31 - 1

i = 0
n = len(s)

# Step 1: Ignore leading spaces
while i < n and s[i] == " ":
    i += 1

# Step 2: Determine the sign
sign = 1

if i < n and s[i] == "-":
    sign = -1
    i += 1
elif i < n and s[i] == "+":
    i += 1

# Step 3: Read consecutive digits
num = 0

while i < n and s[i].isdigit():
    digit = int(s[i])
    num = num * 10 + digit

    # Check overflow while building the number
    if sign == 1 and num > INT_MAX:
        print(INT_MAX)

    if sign == -1 and -num < INT_MIN:
        print(INT_MIN)

    i += 1

# Step 4: Apply sign
sign = sign * num
print(sign)

# Hence its an Integer
print(sign + 1)
```

```python
# Roman to an Integer

s = "MX"

values = {
    'I' : 1,
    'V' : 5,
    'X' : 10,
    'L' : 50,
    'C' : 100,
    'D' : 500,
    'M' : 1000
}

total = 0

print(len(s))

for i in range(len(s)):
    current = values[s[i]]
    print(current)

    # if the current value is smaller we want to subtract it
    if i < len(s) - 1 and current < values[s[i + 1]]:
        total -= current

    else:
        total += current


print("---------------")
print(total)
```

```python
# Integer to an Roman

n = 6

values = [    # Iterating through a dictionary as for value,
    (1000, 'M'),  #  symbol in values: does not work—you need .items().
    (900, 'CM'),
    (500, 'D'),
    (400, 'CD'),
    (100, 'C'),
    (90, 'XC'),
    (50, 'L'),
    (40, 'XL'),
    (10, 'X'),
    (9, 'IX'),
    (5, 'V'),
    (4, 'IV'),
    (1, 'I')
]

result = ""

for value, symbol in values:
    while n >= value :
        result += symbol
        n -= value

print(result)
```
